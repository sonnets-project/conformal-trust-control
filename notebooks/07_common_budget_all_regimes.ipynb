# Common-budget policy and episode records

Train the validation-selected and aggressive policies for each regime,
save their checkpoints and construct counterfactual episode records over
the trust grid. Reloaded checkpoints must reproduce the saved arrays.
Subsequent runs reuse existing checkpoints after checking their hashes.

These records are the inputs to the v1.3 deployment replay. Run from the
experiment directory after the four-regime controller notebook.


```python
from pathlib import Path
import os
import sys
import runpy


def _find_repository_root():
    here = Path.cwd().resolve()
    for candidate in (here, *here.parents):
        if (candidate / 'ctc_runtime.py').is_file():
            return candidate
    raise FileNotFoundError('Start Jupyter from the conformal-trust-control repository.')

KIT_ROOT = _find_repository_root()

RUNNER = KIT_ROOT / 'new-reward' / 'v12r_lineage_autorun.py'
assert RUNNER.exists(), f'missing v1.2R runner: {RUNNER}'
print('Repository:', KIT_ROOT)
print('v1.2R runner:', RUNNER)
runpy.run_path(str(RUNNER), run_name='__main__')
```

## Outputs

The combined panel is saved to
`new-reward/checkpoints/controls_panel_summary_c5_v12r.csv`.
Each regime also produces policy checkpoints, counterfactual ledgers and a
manifest recording the source and input hashes. An incompatible existing
run raises an error; use a new experiment directory for changed inputs.
