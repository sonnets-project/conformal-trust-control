# Deployment-rule comparison

Evaluate six deployment rules across the four test regimes for the aggressive
policy at kappa=2. This produces Table 3. The grid rules use saved
counterfactual ledgers; continuous ACI uses its recorded path with the
warm-up correction. Outputs are written to
`results/published_rules_4regime_v13/`.


```python
from pathlib import Path
import os
import sys
import runpy


def _find_repository_root():
    here = Path.cwd().resolve()
    for candidate in (here, *here.parents):
        if (candidate / 'ctc_runtime.py').is_file():
            return candidate
    raise FileNotFoundError('Start Jupyter from the conformal-trust-control repository.')

KIT_ROOT = _find_repository_root()

print('Repository:', KIT_ROOT)
script = KIT_ROOT / 'new-reward' / 'published_rules_4regime_replay.py'
namespace = runpy.run_path(str(script))
paths = namespace['run'](KIT_ROOT)

import pandas as pd
from IPython.display import display, Markdown, Image

display(Markdown('## 24 regime-rule results'))
panel = pd.read_csv(paths['panel_csv'])
display(panel[['regime', 'display_name', 'mean_dpnl_bp', 'breach',
               'mean_lambda', 'budget_mean_bp']])
display(Markdown('## Best feasible published-method-inspired baseline'))
display(pd.read_csv(paths['best_csv']))
if paths['plot_png'].exists():
    display(Markdown('## Cross-regime trade-off figure'))
    display(Image(filename=str(paths['plot_png'])))
display(Markdown(
    '## Completed\n'
    f"Outputs: `{paths['output_dir']}`\n\n"
    f"Manifest: `{paths['manifest'].name}`"
))
```
