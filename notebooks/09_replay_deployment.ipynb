# Deployment replay

Apply the paper's controller protocol to the saved counterfactual ledgers:

- Use the Black–Scholes fallback until 15 completed episodes are available.
- Count breaches over all test episodes, including warm-up.
- Process completion feedback in batches at each decision date.

Run after the common-budget policy and episode records have been generated.
The results supply Tables 4–5 and the breach-rate uncertainty analysis.


```python
from pathlib import Path
import os
import sys
import runpy


def _find_repository_root():
    here = Path.cwd().resolve()
    for candidate in (here, *here.parents):
        if (candidate / 'ctc_runtime.py').is_file():
            return candidate
    raise FileNotFoundError('Start Jupyter from the conformal-trust-control repository.')

KIT_ROOT = _find_repository_root()

print('Repository:', KIT_ROOT)
script = KIT_ROOT / 'new-reward' / 'warm0_batch_replay.py'
namespace = runpy.run_path(str(script))
namespace['run'](KIT_ROOT)
```
